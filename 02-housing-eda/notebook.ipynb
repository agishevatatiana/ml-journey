```python
import pandas as pd

file_path = '../external-data/train.csv'

housing_df = pd.read_csv(file_path)

# check the size of data frame, x - rows number, y - columns number
housing_df.shape
```

```python
# select first 3 house prices using index-based approach, SalePrice is the last 81st column
housing_df.iloc[:3, 80]
```

```python
# select SalePrice, LotArea and YearBuilt of the last 3 houses using label-based selection
housing_df.loc[1457:, ['SalePrice', 'LotArea', 'YearBuilt']]
```

```python
# set index to the SalePrice
housing_df.set_index("SalePrice")
```

```python
# creating the Series of True/False values (==)
condition = housing_df.SaleCondition == 'Abnorml'

# select only the rows with True, SaleCondition is 'Abnorml', using loc
housing_df.loc[condition]
```

```python
# complex condition to select Abnormal with area smaller than 6200
complexCondition = (housing_df.SaleCondition == 'Abnorml') & (housing_df.LotArea < 6200)

# select the rows with this houses
housing_df.loc[complexCondition]
```

```python
# look for Abnormal or Partial
housing_df.loc[housing_df.SaleCondition.isin(['Abnormal', 'Partial'])]
```

```python
# look for houses with the pool
housing_df.loc[housing_df.PoolQC.notnull()]
```

```python
# add 'PricePerSqM' column to df
housing_df['PricePerSqM'] = housing_df.SalePrice / housing_df.LotArea

housing_df.head()
```

```python
# describe data of PricePerSqM column
housing_df.PricePerSqM.describe()
```

```python
# check only mean value of PricePerSqM
housing_df.PricePerSqM.mean()
```

```python
# check unique values and number
unv = housing_df.PricePerSqM.unique()
unv, unv.size
```

```python
# unique values and how often they occur
housing_df.PricePerSqM.value_counts()
```

```python
# check the PricePerSqM delta from mean value (remean), also map() or apply() can be used
PricePerSqM_mean = housing_df.PricePerSqM.mean()
housing_df.PricePerSqM - PricePerSqM_mean
```

```python
# get the most expensive house in every SaleCondition
housing_df.groupby('SaleCondition').SalePrice.max()
```

```python
# get the first house in the df of every SaleCondition
housing_df.groupby('SaleCondition').apply(lambda df: df.iloc[0])
```

```python
# SalePrice statistics by SaleCondition
housing_df.groupby(['SaleCondition']).SalePrice.agg([len, min, max])
```

```python
# multi-indexes
sc_y_indexed = housing_df.groupby(['SaleCondition', 'YearBuilt']).SalePrice.agg([len])
sc_y_indexed
```

```python
# get back to regular indexing
sc_y_indexed.reset_index()
```

```python
# sort by len
sc_y_indexed.sort_values(by='len')
```

```python
# desc order
sc_y_indexed.sort_values(by='len', ascending=False)
```

```python
# sort by index
sc_y_indexed = sc_y_indexed.reset_index()
sc_y_indexed.sort_index()
```

```python
# sortby several columns
sc_y_indexed.sort_values(by=['len', 'YearBuilt'], ascending=False)
```

```python
# select houses without pool
housing_df.loc[pd.isnull(housing_df.PoolQC)]
```

```python
# fill NaN with some value
housing_df.PoolQC = housing_df.PoolQC.fillna("No Pool")
housing_df.head()
```

```python
# replace some values with another
housing_df.LotShape = housing_df.LotShape.replace("Reg", "regtangle")
housing_df.head()
```

```python
# renaming column
housing_df.rename(columns={'SalePrice': 'Price'})
```

```python
# rename index (manual alternative to set_index())
housing_df.rename(index={0: 'firstEntry', 1: 'secondEntry'})
```

```python
# change row and column names
housing_df.rename_axis("test0", axis='rows').rename_axis("test", axis='columns')
```

```python
# combining data vertically
test_file_path = '../external-data/test.csv'
housing_df1 = pd.read_csv(test_file_path)

pd.concat([housing_df, housing_df1])
```

```python
# combine data horizontaly (join)
left = housing_df.set_index(['Id', 'LotArea'])
right = housing_df1.set_index(['Id', 'LotArea'])

left.join(right, lsuffix='_train', rsuffix='_test')
```
